```python
#punto 1)
def max_heapify(heap, n, i):
    mayor = i
    izq = 2 * i + 1
    der = 2 * i + 2

    # Comparar con el hijo izquierdo (la prioridad está en la posición [2])
    if izq < n and heap[izq][2] > heap[mayor][2]:
        mayor = izq

    # Comparar con el hijo derecho
    if der < n and heap[der][2] > heap[mayor][2]:
        mayor = der

    # Si el mayor no es el padre, se intercambian y se sigue ajustando hacia abajo
    if mayor != i:
        temp = heap[i]
        heap[i] = heap[mayor]
        heap[mayor] = temp
        max_heapify(heap, n, mayor)

def build_max_heap(heap):
    n = len(heap)
    # Se recorre desde el último nodo padre hasta la raíz (índice 0)
    i = (n // 2) - 1
    while i >= 0:
        max_heapify(heap, n, i)
        i = i - 1

def mostrar_arreglo(heap, titulo):
    print("\n" + titulo)
    i = 0
    while i < len(heap):
        print("  [" + str(i) + "] Código:", heap[i][0], "| Tarea:", heap[i][1], "| Prioridad:", heap[i][2])
        i = i + 1

def mostrar_arbol(heap):
    print("\nÁrbol correspondiente al Max-Heap:")
    n = len(heap)
    nivel = 0
    i = 0
    while i < n:
        nodos_en_nivel = 2 ** nivel
        linea = "  Nivel " + str(nivel) + ": "
        contador = 0
        while contador < nodos_en_nivel and i < n:
            linea = linea + "[" + heap[i][0] + " (" + str(heap[i][2]) + ")]  "
            i = i + 1
            contador = contador + 1
        print(linea)
        nivel = nivel + 1

# Ingreso de datos por teclado
tareas = []
cantidad_tareas = int(input("¿Cuántas tareas desea ingresar?: "))

i = 0
while i < cantidad_tareas:
    print("\n--- Tarea " + str(i + 1) + " ---")
    codigo = input("Código (ej. T01): ").strip()
    nombre = input("Descripción de la tarea: ").strip()
    prioridad = int(input("Prioridad (número entero): "))
    tareas.append([codigo, nombre, prioridad])
    i = i + 1

# Mostrar arreglo inicial, construir Max-Heap y mostrar resultados
mostrar_arreglo(tareas, "Arreglo inicial:")
build_max_heap(tareas)
mostrar_arreglo(tareas, "Max-Heap resultante:")
mostrar_arbol(tareas)


punto 2
def extract_max(heap):
    n = len(heap)
    if n == 0:
        return None

    tarea_max = heap[0]
    heap[0] = heap[n - 1]
    heap.pop()

    if len(heap) > 0:
        max_heapify(heap, len(heap), 0)

    return tarea_max

def procesar_dos_mayores(heap):
    paso = 1
    while paso <= 2:
        if len(heap) > 0:
            procesada = extract_max(heap)
            print("\n>>> Extracción #" + str(paso) + " - Tarea procesada:", procesada[0], "-", procesada[1], "(Prioridad:", procesada[2], ")")
            mostrar_arreglo(heap, "Heap resultante tras extracción #" + str(paso) + ":")
        else:
            print("\nNo hay más tareas en el Heap.")
        paso = paso + 1

def pedir_nueva_tarea():
    print("\n--- Ingresar nueva tarea ---")
    codigo = input("Código de la nueva tarea: ").strip()
    nombre = input("Descripción de la nueva tarea: ").strip()
    prioridad = int(input("Prioridad de la nueva tarea: "))
    return [codigo, nombre, prioridad]

def subir_en_heap(heap, indice):
    while indice > 0:
        padre = (indice - 1) // 2
        if heap[indice][2] > heap[padre][2]:
            intercambiar(heap, indice, padre)
            indice = padre
        else:
            break

def insertar_tarea(heap, nueva_tarea):
    heap.append(nueva_tarea)
    ultimo_indice = len(heap) - 1
    subir_en_heap(heap, ultimo_indice)

def ejecutar_punto_2():
    procesar_dos_mayores(tareas)
    nueva_tarea = pedir_nueva_tarea()
    insertar_tarea(tareas, nueva_tarea)
    mostrar_arreglo(tareas, "Heap después de la inserción:")
    mostrar_arbol(tareas)

def main():
    ejecutar_punto_2()

main()

punto 3

import math
import matplotlib.pyplot as plt

dispositivos = []
aristas = []

def pedir_dispositivos(lista_dispositivos):
    cantidad = int(input("¿Cuántos dispositivos desea ingresar?: "))
    i = 0
    while i < cantidad:
        print("\n--- Dispositivo " + str(i + 1) + " ---")
        nombre = input("Nombre del dispositivo (ej. Servidor): ").strip()
        id_disp = input("ID del dispositivo (ej. A): ").strip()
        x = float(input("Coordenada X: "))
        y = float(input("Coordenada Y: "))
        lista_dispositivos.append([nombre, id_disp, x, y])
        i = i + 1

def graficar_dispositivos(lista_dispositivos):
    plt.figure(figsize=(8, 5))
    i = 0
    while i < len(lista_dispositivos):
        nombre = lista_dispositivos[i][0]
        id_d = lista_dispositivos[i][1]
        x = lista_dispositivos[i][2]
        y = lista_dispositivos[i][3]
        plt.scatter(x, y, color="blue", s=100)
        plt.text(x + 0.2, y + 0.2, id_d + " (" + nombre + ")")
        i = i + 1

    plt.title("Posición de los dispositivos")
    plt.xlabel("Coordenada X")
    plt.ylabel("Coordenada Y")
    plt.grid(True)
    plt.show()

def calcular_distancia(x1, y1, x2, y2):
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

def generar_aristas(lista_dispositivos, lista_aristas):
    n = len(lista_dispositivos)
    i = 0
    while i < n:
        j = i + 1
        while j < n:
            distancia = calcular_distancia(
                lista_dispositivos[i][2], lista_dispositivos[i][3],
                lista_dispositivos[j][2], lista_dispositivos[j][3]
            )
            lista_aristas.append([lista_dispositivos[i][1], lista_dispositivos[j][1], distancia])
            j = j + 1
        i = i + 1

def mostrar_aristas(lista_aristas):
    print("\nConexiones posibles y sus distancias (" + str(len(lista_aristas)) + " aristas):")
    i = 0
    while i < len(lista_aristas):
        print(" ", lista_aristas[i][0] + " - " + lista_aristas[i][1] + ": " + ("%.2f" % lista_aristas[i][2]))
        i = i + 1

def ejecutar_punto_3():
    pedir_dispositivos(dispositivos)
    graficar_dispositivos(dispositivos)
    generar_aristas(dispositivos, aristas)
    mostrar_aristas(aristas)

def main():
    ejecutar_punto_3()

main()


punto 4


mst = []

def ordenar_aristas(lista_aristas):
    m = len(lista_aristas)
    i = 0
    while i < m:
        j = 0
        while j < m - i - 1:
            if lista_aristas[j][2] > lista_aristas[j + 1][2]:
                intercambiar(lista_aristas, j, j + 1)
            j = j + 1
        i = i + 1

def crear_grupos(lista_dispositivos):
    grupos = {}
    i = 0
    while i < len(lista_dispositivos):
        id_d = lista_dispositivos[i][1]
        grupos[id_d] = id_d
        i = i + 1
    return grupos

def buscar_grupo(grupos, nodo):
    while grupos[nodo] != nodo:
        nodo = grupos[nodo]
    return nodo

def unir_grupos(grupos, nodo1, nodo2):
    raiz1 = buscar_grupo(grupos, nodo1)
    raiz2 = buscar_grupo(grupos, nodo2)
    if raiz1 != raiz2:
        grupos[raiz2] = raiz1

def kruskal(lista_dispositivos, lista_aristas, lista_mst):
    ordenar_aristas(lista_aristas)
    grupos = crear_grupos(lista_dispositivos)
    conexiones_necesarias = len(lista_dispositivos) - 1

    i = 0
    while i < len(lista_aristas):
        origen = lista_aristas[i][0]
        destino = lista_aristas[i][1]
        peso = lista_aristas[i][2]

        if buscar_grupo(grupos, origen) != buscar_grupo(grupos, destino):
            unir_grupos(grupos, origen, destino)
            lista_mst.append([origen, destino, peso])

        if len(lista_mst) == conexiones_necesarias:
            break

        i = i + 1

def calcular_longitud_total(lista_mst):
    total = 0.0
    i = 0
    while i < len(lista_mst):
        total = total + lista_mst[i][2]
        i = i + 1
    return total

def mostrar_mst(lista_mst, total):
    print("Conexiones seleccionadas (MST):")
    i = 0
    while i < len(lista_mst):
        print(lista_mst[i][0] + " - " + lista_mst[i][1] + ": " + ("%.2f" % lista_mst[i][2]))
        i = i + 1
    print("Longitud total de cable: " + ("%.2f" % total))

def ejecutar_punto_4():
    kruskal(dispositivos, aristas, mst)
    total = calcular_longitud_total(mst)
    mostrar_mst(mst, total)

def main():
    ejecutar_punto_4()

main()

punto 5

def obtener_diccionarios(lista_dispositivos):
    coord_x = {}
    coord_y = {}
    nombres = {}
    i = 0
    while i < len(lista_dispositivos):
        id_d = lista_dispositivos[i][1]
        nombres[id_d] = lista_dispositivos[i][0]
        coord_x[id_d] = lista_dispositivos[i][2]
        coord_y[id_d] = lista_dispositivos[i][3]
        i = i + 1
    return coord_x, coord_y, nombres

def dibujar_conexiones_mst(lista_mst, coord_x, coord_y):
    i = 0
    while i < len(lista_mst):
        u = lista_mst[i][0]
        v = lista_mst[i][1]
        peso = lista_mst[i][2]

        x1 = coord_x[u]
        y1 = coord_y[u]
        x2 = coord_x[v]
        y2 = coord_y[v]

        plt.plot([x1, x2], [y1, y2], color="green", linewidth=2)

        xm = (x1 + x2) / 2
        ym = (y1 + y2) / 2
        plt.text(xm, ym + 0.15, "%.2f" % peso, color="darkred", fontweight="bold")
        i = i + 1

def dibujar_dispositivos(lista_dispositivos, nombres):
    i = 0
    while i < len(lista_dispositivos):
        id_d = lista_dispositivos[i][1]
        x = lista_dispositivos[i][2]
        y = lista_dispositivos[i][3]
        plt.scatter(x, y, color="blue", s=120, zorder=5)
        plt.text(x + 0.2, y + 0.2, id_d + " (" + nombres[id_d] + ")", fontweight="bold")
        i = i + 1

def graficar_mst_final(lista_dispositivos, lista_mst):
    coord_x, coord_y, nombres = obtener_diccionarios(lista_dispositivos)
    total = calcular_longitud_total(lista_mst)

    plt.figure(figsize=(9, 6))
    dibujar_conexiones_mst(lista_mst, coord_x, coord_y)
    dibujar_dispositivos(lista_dispositivos, nombres)
    plt.title("Árbol de Expansión Mínima (MST) - Longitud total de cable: " + ("%.2f" % total))
    plt.xlabel("Coordenada X")
    plt.ylabel("Coordenada Y")
    plt.grid(True)
    plt.show()

def ejecutar_punto_5():
    graficar_mst_final(dispositivos, mst)

def main():
    ejecutar_punto_5()

main()
```
